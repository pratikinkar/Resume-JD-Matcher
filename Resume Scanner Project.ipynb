```python
# Importing required Libraries 
import os 
from PyPDF2 import PdfReader
```

```python
def extract_text_from_pdf(pdf_path):
    text = ""
    reader = PdfReader(pdf_path)
    for page in reader.pages:
        text += page.extract_text()
    return text
```

```python
folder = "resumes/"
resumes = {}
for file in os.listdir(folder):
    if file.endswith(".pdf"):
        resumes[file] = extract_text_from_pdf(os.path.join(folder, file))
```

```python
# Importing Libraries to process on extract text 
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

```python
def preprocess_text(text):
    text = re.sub(r'[^a-zA-Z]','',text)
    text = text.lower()
    tokens = text.split()
    tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return " ".join(tokens)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

jd = input('Enter a job Description:')

documents = [preprocess_text(jd)] + [
    preprocess_text(text) for text in resumes.values()
]

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(documents)

similarity_scores = cosine_similarity(vectors[0:1], vectors[1:]).flatten()

for i, (filename, score) in enumerate(zip(resumes.keys(), similarity_scores)): 
    print(f"{filename}: {score:0.2f}")
```
